```python

# CODOMAX INTERNSHIP
# MODULE 02 - DAY 6
# PROMPT ENGINEERING BASICS

!pip install transformers sentencepiece -q

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Step 1: Load AI Model
model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("AI model loaded successfully!")

# Step 2: Create Prompt Function
def generate_response(prompt, max_new_tokens=100):
    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens
    )
    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )
    return answer

print("Prompt function created successfully!")

# Step 3: Zero-Shot Prompting
zero_shot_prompt = "Explain Generative AI in simple words."

print("\n===== ZERO-SHOT PROMPTING =====")
print("Prompt:", zero_shot_prompt)
print("AI Response:", generate_response(zero_shot_prompt))

# Step 4: Few-Shot Prompting
few_shot_prompt = """
Classify the sentiment as Positive or Negative.

Example:
Text: I love this product.
Sentiment: Positive

Text: The service is very bad.
Sentiment: Negative

Text: This course is very useful.
Sentiment:
"""

print("\n===== FEW-SHOT PROMPTING =====")
print("AI Response:", generate_response(few_shot_prompt))

# Step 5: Role-Based Prompting
role_prompt = """
Role: You are a Python programming teacher.
Context: A beginner student is learning Python.
Instruction: Explain what a Python variable is.
Constraint: Use simple English and give one example.
"""

print("\n===== ROLE-BASED PROMPTING =====")
print("AI Response:", generate_response(role_prompt))

# Step 6: Day 6 Conclusion
print("\n===== DAY 6 CONCLUSION =====")
print("1. Learned zero-shot prompting.")
print("2. Practised few-shot prompting.")
print("3. Explored role-based prompting.")
print("4. Used context, instructions and constraints.")
print("DAY 6 COMPLETED SUCCESSFULLY!")
```

```python

# DAY 8 - PROMPT TEMPLATES AND PROMPT CHAINING

!pip install transformers sentencepiece -q

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load AI Model
model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Function to generate AI responses
def generate_response(prompt):
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

# --------------------------------
# 1. PROMPT TEMPLATE
# --------------------------------

print("DAY 8 - PROMPT TEMPLATES")

template = "Explain {topic} in simple words."

topic = "Artificial Intelligence"

prompt = template.format(topic=topic)

response = generate_response(prompt)

print("Topic:", topic)
print("AI Response:", response)


# --------------------------------
# 2. REUSABLE PROMPT TEMPLATE
# --------------------------------

print("\nREUSABLE PROMPT TEMPLATE")

def explain_topic(topic):
    prompt = f"Explain {topic} in three simple sentences."
    return generate_response(prompt)

print(explain_topic("Machine Learning"))
print(explain_topic("Deep Learning"))


# --------------------------------
# 3. PROMPT CHAINING
# --------------------------------

print("\nPROMPT CHAINING")

# Step 1: Generate explanation
topic = "Generative AI"

prompt1 = f"Explain {topic} in simple words."

explanation = generate_response(prompt1)

print("Step 1 - Explanation:")
print(explanation)

# Step 2: Summarize the explanation
prompt2 = f"Summarize this text in one short sentence: {explanation}"

summary = generate_response(prompt2)

print("\nStep 2 - Summary:")
print(summary)


# --------------------------------
# 4. FINAL OUTPUT
# --------------------------------

print("\nFINAL OUTPUT")
print("Original Topic:", topic)
print("AI Explanation:", explanation)
print("AI Summary:", summary)

print("\nDAY 8 COMPLETED SUCCESSFULLY!")
```

```python

# DAY 9 - INTRODUCTION TO LANGCHAIN

!pip install -q langchain langchain-core

# Import required libraries
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableLambda

print("DAY 9 - INTRODUCTION TO LANGCHAIN")

# ---------------------------------
# 1. CREATE A PROMPT TEMPLATE
# ---------------------------------

template = """
Explain {topic} in simple words.
Give two real-life examples.
"""

prompt = PromptTemplate.from_template(template)

# ---------------------------------
# 2. CREATE A SIMPLE AI RESPONSE
# ---------------------------------

def generate_answer(text):
    return (
        "Artificial Intelligence is a technology "
        "that helps machines perform intelligent tasks. "
        "Examples: Voice assistants and recommendation systems."
        if "Artificial Intelligence" in text
        else "Machine Learning allows computers to learn "
             "from data and make predictions."
    )

# ---------------------------------
# 3. CREATE A LANGCHAIN CHAIN
# ---------------------------------

model = RunnableLambda(generate_answer)

chain = prompt | model

# ---------------------------------
# 4. RUN THE CHAIN
# ---------------------------------

result = chain.invoke({
    "topic": "Artificial Intelligence"
})

print("\nTopic: Artificial Intelligence")
print("\nAI Response:")
print(result)

# ---------------------------------
# 5. TRY ANOTHER TOPIC
# ---------------------------------

result2 = chain.invoke({
    "topic": "Machine Learning"
})

print("\nTopic: Machine Learning")
print("\nAI Response:")
print(result2)

print("\nDAY 9 COMPLETED SUCCESSFULLY!")
```

```python

# DAY 10 - BUILDING A SIMPLE AI CHATBOT

print("WELCOME TO MY AI CHATBOT!")
print("Type 'exit' to stop the chatbot.")

# Chatbot function
def chatbot(user_input):

    user_input = user_input.lower()

    if "hello" in user_input or "hi" in user_input:
        return "Hello! How can I help you?"

    elif "name" in user_input:
        return "I am your Python AI Chatbot."

    elif "artificial intelligence" in user_input:
        return "AI enables machines to perform intelligent tasks."

    elif "machine learning" in user_input:
        return "Machine Learning helps computers learn from data."

    elif "python" in user_input:
        return "Python is a popular programming language."

    elif "thank" in user_input:
        return "You're welcome!"

    elif "bye" in user_input:
        return "Goodbye! Have a nice day!"

    else:
        return "Sorry, I don't understand. Try another question."


# Chatbot interaction
while True:

    user = input("\nYou: ")

    if user.lower() == "exit":
        print("Chatbot: Thank you for chatting!")
        break

    response = chatbot(user)

    print("Chatbot:", response)


print("\nDAY 10 COMPLETED SUCCESSFULLY!")
```
